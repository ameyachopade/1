## 1. Action Queue Intake & Distribution

Loading the prioritized decay queue and inspecting the initial distribution of recommended actions.

```python
import pandas as pd
import numpy as np

# Load baseline outputs
df_queue = pd.read_csv('work/outputs/baseline_action_score.csv')
raw = pd.read_csv('data/raw/content_refresh_anonymized.csv')
df_queue['is_declining_label'] = (raw['trend_direction'] == 'down').astype(int)

print(f"Action Queue Loaded: {len(df_queue):,} items")
print("Action Label Distribution:")
print(df_queue['action_label'].value_counts(normalize=True).round(3))
```

## 2. Action Tiers & Decay Density

Verifying that prioritized action tiers carry elevated organic decay concentrations compared to passive monitoring.

```python
# Action Tiers & Thresholding
t1 = df_queue[df_queue['action_label'] == 'review_snippet_and_ctr']
t2 = df_queue[df_queue['action_label'] == 'content_refresh']
t3 = df_queue[df_queue['action_label'] == 'monitor']

print(f"Tier 1 (Snippet/CTR Review): {len(t1):,} pages | Declining Rate: {t1['is_declining_label'].mean():.1%}")
print(f"Tier 2 (Content Refresh):    {len(t2):,} pages | Declining Rate: {t2['is_declining_label'].mean():.1%}")
print(f"Tier 3 (Passive Monitoring): {len(t3):,} pages | Declining Rate: {t3['is_declining_label'].mean():.1%}")
```

## 3. Reason Code Attribution

Examining human-interpretable reason codes that justify why each page was selected.

```python
# Reason Code Breakdown for Actionable Items
actionable = df_queue[df_queue['action_label'] != 'monitor']
print("Reason Codes for Prioritized Queue:")
print(actionable['reason_code'].value_counts())
```

## 4. Playbook Top Picks Inspection

Reviewing the top ranked recommendations ready for human editorial review.

```python
# Playbook Recommendations Sample
top_picks = df_queue.head(5)[[
    'baseline_rank', 'content_id', 'client_id', 'action_label',
    'reason_code', 'impressions_90d', 'avg_position', 'ctr', 'days_since_last_update'
]]
print("Top 5 Action Recommendations:")
print(top_picks.to_string(index=False))
```

## 5. Editorial Execution Guidelines

Standard operating procedures for content teams acting on weekly priority batches.

```python
# Operational Summary
print("Editorial Playbook Summary:")
print("1. review_snippet_and_ctr: Target pages with high visibility and poor click capture.")
print("2. content_refresh: Target pages with aging content in striking distance.")
print("3. monitor: Protect evergreen articles from re-indexing volatility.")
```
