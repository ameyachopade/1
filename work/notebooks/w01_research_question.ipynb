## 1. My lane (or freestyle) and why

**Selected Lane:** **Refresh / Content Opportunity Scoring**

Managing enterprise content libraries requires deciding which aging pages warrant editorial updates and which should be left alone. Manual inspection does not scale across tens of thousands of URLs, while scheduled arbitrary refreshes squander editorial time. By scoring content decay risk using pre-trend search metrics, we build an automated queue to defend slipping high-value keywords before traffic collapses.

```python
import pandas as pd
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
df['is_declining_label'] = (df['trend_direction'] == 'down').astype(int)
print(f"Selected Lane: Refresh / Content Opportunity Scoring")
print(f"Total Dataset Items: {len(df):,} rows across {df['client_id'].nunique()} clients")
```

## 2. The question: decision, action, cost of a wrong call

- **Decision Supported:** Which content pages should the editorial team prioritize for immediate content refreshes or meta/snippet optimization this week?
- **Who Acts on It:** Editorial content managers and SEO specialists.
- **Cost of a False Positive:** 2–4 hours of editorial time wasted editing a page that was already stable.
- **Cost of a False Negative:** High-intent commercial content slips from Page 1 to Page 3+, losing organic traffic that takes months of backlinking to regain.

```python
# Decision & Unit of Analysis
print("Unit of Analysis: Single web page over a trailing 90-day window")
print("Target Outcome: is_declining_label (1 if traffic trend is down, 0 otherwise)")
print(f"Baseline Population Declining Rate: {df['is_declining_label'].mean():.1%}")
```

## 3. Quick look at the data (2-3 real numbers)

We inspect the starter dataset to verify the opportunity size:
- Over 31% of the catalog has gone untouched for over 90 days.
- Over 63% of pages sit in striking distance (positions 4–20) where ranking decay actively risks losing Page 1 placement.
- The base decline rate is 54.2%, confirming a high-frequency business problem.

```python
# Quick look at real data numbers
stale_pages = (df['days_since_last_update'] >= 90).sum()
striking_pages = ((df['avg_position'] >= 4) & (df['avg_position'] <= 20)).sum()
print(f"Pages Stale >= 90 days: {stale_pages:,} ({stale_pages/len(df):.1%})")
print(f"Pages in Striking Distance (Pos 4-20): {striking_pages:,} ({striking_pages/len(df):.1%})")
print(f"Median 90-Day Impressions: {df['impressions_90d'].median():.0f}")
```
